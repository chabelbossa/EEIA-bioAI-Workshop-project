# Test de connexion Colab + Google Drive

Ce notebook ne charge pas les embeddings en mémoire. Il vérifie seulement que le code s'exécute sur Colab et que les trois fichiers Evo2 sont visibles sur Drive.

```python
import platform
import sys

IN_COLAB = "google.colab" in sys.modules
print("Runtime Colab :", IN_COLAB)
print("Python :", platform.python_version())
print("Plateforme :", platform.platform())
assert IN_COLAB, "Sélectionnez d'abord le noyau Colab dans VS Code."
```

```python
from google.colab import drive

drive.mount("/content/drive")
```

```python
from pathlib import Path

EMBEDDINGS_DIR = Path(
    "/content/drive/MyDrive/EEIA-bioAI-Workshop-project/2-data/embeddings"
)
required = [EMBEDDINGS_DIR / f"{split}.npz" for split in ("train", "val", "test")]
missing = [path for path in required if not path.is_file()]

print("Dossier vérifié :", EMBEDDINGS_DIR)
for path in required:
    if path.is_file():
        print(f"OK  {path.name}: {path.stat().st_size / 1024**3:.2f} Gio")
    else:
        print(f"MANQUANT  {path.name}")

assert not missing, (
    "Placez train.npz, val.npz et test.npz dans le dossier Drive affiché ci-dessus, "
    "ou modifiez EMBEDDINGS_DIR."
)
```
